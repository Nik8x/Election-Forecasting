# Feature engineering

Imputing the missing Rasmussen and SurveyUSA values (the original used
R's `mice` package for multiple imputation), and building the reduced,
low-collinearity feature set notebook 01 pointed to.

```python
import os
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = pd.read_csv('PollingData.csv')

impute_cols = ['Rasmussen', 'SurveyUSA', 'PropR', 'DiffCount']
imputer = IterativeImputer(random_state = 42)
imputed = imputer.fit_transform(df[impute_cols])
df[['Rasmussen', 'SurveyUSA']] = pd.DataFrame(imputed, columns = impute_cols)[['Rasmussen', 'SurveyUSA']]

df.isnull().sum()
```

`IterativeImputer` (scikit-learn's equivalent to R's `mice`, modeling each column from the others rather than just filling in a mean) fills in the gaps using `PropR` and `DiffCount`, which are complete, plus whichever of Rasmussen/SurveyUSA is available for that row.

```python
train = df[df.Year.isin([2004, 2008])].reset_index(drop = True)
test = df[df.Year == 2012].reset_index(drop = True)

import pickle
os.makedirs('data', exist_ok = True)
with open('data/model_data.pkl', 'wb') as f:
    pickle.dump({'train': train, 'test': test}, f)
train.shape, test.shape
```
